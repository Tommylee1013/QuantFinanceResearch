## Causal Factor Analysis is a necessary condition for Investment Efficiency

### Abstract

이 논문은 포트폴리오 최적화 맥락에서 팩터 모형의 잘못된 명세가 초래하는 심각한 결과를 드러낸다. 우리는 인과적 팩터 모형 없이 효율적 경계를 추정하는 것이 불가능하며, 기존의 (연관적) 팩터 모형 패러다임이 투자 비효율성을 초래한다는 것을 보여준다. 또한 팩터 모형의 잘못된 명세로 인해 발생하는 편향이 매우 커서, 투자자가 평균과 공분산을 완벽하게 알고 있는 경우에도 매도해야 할 자산을 매수하고 매수해야 할 자산을 매도하는 포트폴리오를 구성하게 될 수 있음을 보여준다. 우리의 연구 결과는 현재 자산 운용 산업에서 사용되는 포트폴리오 구성 패러다임의 과학적 타당성에 의문을 제기한다. 이러한 문제를 극복하기 위해 학계와 실무자는 금융 경제학 문헌을 보다 과학적으로 엄밀한 인과적 팩터 투자라는 기반 위에 다시 세워야 한다.

### Introduction

현대 포트폴리오 이론의 중요한 목표는 단위 위험당 기대 수익을 최대화하는 효율적인 최적 포트폴리오를 계산하는 것이다. 위험은 일반적으로 두 가지 상호 보완적인 방식으로 정의된다: 

1. 기대 포트폴리오 수익률의 표준편차로 정의하는 방식 (예: Markowitz [1952])
2. 크기, 가치, 성장, 모멘텀, 품질, 수익률, 저변동성, 캐리, 유동성, 거시경제적 또는 고유 요인과 같은 요인의 조합에 대한 포트폴리오의 노출로 정의하는 방식 (예: Campbell et al. [1997])이다.

2번 방식의 정의는 항상 전체 위험을 "factor"로 알려진 원천으로 분류하는 것을 포함하며, 1번 방식의 정의는 일반적으로 요인 기반 공분산 행렬을 구축하는 것을 포함한다. 따라서 투자자가 1번 또는 2번 또는 두 가지 방식을 모두 사용하여 위험을 통제하더라도, 최적 포트폴리오를 추정하기 위한 표준 접근법은 요인 모형의 사용을 포함한다.

요인 투자 문헌에서는 최소 편향의 비편향 추정량(MVUE, Minimum Variance Unbiased Estimator)을 통해 노출과 위험 프리미엄을 계산하며, Least Squares Method가 그 대표적인 예이다. 관련 연구로는 Fama와 MacBeth [1973], Fama와 French [1993, 1996, 2015], Carhart [1997], Ang et al. (2006), Qian et al. [2007], Ferson [2019] 등이 있다.

Gauss-Markov 정리에 따르면, 추정량이 비편향이 되기 위해서는 모형이 올바르게 명시되어야 한다. 모형이 올바르게 명시되려면 연구자가 요인과 결과 사이의 관계에서 존재하는 모든 backdoor path를 차단하는 변수 집합(이를 "backdoor adjustment set"이라고 함)을 신중하게 선택해야 한다 (Pearl [2009]). 올바른 backdoor adjustment set을 선택하려면 인과 메커니즘에 관여하는 변수 간 상호 의존 구조를 정의하는 인과 그래프에 대한 지식이 필요하다.

López de Prado [2023a, 2023b, 2023c]는 요인 투자 문헌의 중요한 결함이 요인을 연관성(associational) 용어로 정의한다는 점이라고 지적하였다. 예를 들어, Fama와 French [2015]는 기존의 3요인 모형에 두 가지 추가 요인을 포함하면서 "5요인 모형이 평균 수익을 설명하는 데 있어 3요인 모형보다 더 우수하다"는 논리를 제시하였다. 이러한 상관 기반(즉, 연관적) 논리는 비편향 추정량을 사용하는 것과 일관되지 않으며, 비편향 추정량을 사용하려면 인과 그래프를 통해 올바르게 명시된 모형을 선택해야 한다 (López de Prado와 Zoonekynd [2024], López de Prado et al. [2024] 참조).

이 글은 포트폴리오 최적화 맥락에서 Factor Model의 오류가 초래하는 심각한 결과를 연구한다. 우리는 인과적 Factor Model이 투자 효율성의 필요 조건임을 보여주며, 기존의 연관적 Factor Model의 패러다임이 투자 비효율성을 초래함을 밝힌다.

### Factor Model

$N$개의 자산이 있고, 각각의 수익률을 $X_1, \dots, X_N$으로 나타낸다. 또한 $K$개의 위험 요인이 있으며, 이들의 수익률을 $F_1, \dots, F_K$로 나타낸다. 이들 간의 관계는 다음과 같은 방정식으로 표현된다

$$X = BF + \varepsilon$$

여기서 $B \in \mathbb{R}^{N \times K}$는 exposure matrix로, 자산이 위험 요인에 노출된 정도를 나타내며, $\varepsilon \in \mathbb{R}^N$는 asset-specific returns를 의미한다. 자산 수익률 $X$의 분산-공분산 행렬은 다음과 같이 표현할 수 있다.

$$
\begin{align}
V = \text{Var}[X] =& B\Sigma B^\prime + \Delta \\
\Sigma =& \text{Var}[F] \\
\Delta =& \text{Var}[\varepsilon]
\end{align}
$$

### Investment Strategy

투자자가 특정 요인 노출을 달성하면서 분산을 최소화하는 최적 포트폴리오 $\omega$를 계산하고자 한다고 가정하자. 이를 형식적으로 표현하면 다음과 같다.

$$\min_\omega ~~ \omega^\prime V \omega \text{  such that :  } \omega^\prime B = c$$

여기서 $\omega$는 크기가 $N\times 1$인 벡터, $c$는 크기가 $1 \times K$인 벡터, $V$는 $N \times N$의 양의 정부호(positive-definite) 분산-공분산 행렬이다. 위 최적화 문제의 closed-form solution은 다음과 같다.

$$
\begin{align}
\omega^* =& V^{-1}BA^{-1}c^\prime \\
A =& B^\prime V^{-1} B
\end{align}
$$

특히, $N=K$인 경우 해는 다음과 같이 단순화된다.

$$\omega^* = B^{\prime -1}c^\prime$$

한 예로, 일반성을 잃지 않고, 두 자산과 두 요인(즉, $N=2, K=2$)의 경우를 고려하자. 이 경우, 투자자는 $F_2$에 대해 unit exposure을 가지면서 $F_1$에 대한 노출을 헤지하고자 한다. 따라서, $c=[0,1]$이다. 또한, 추가적인 분석을 위해 $K=1,N=2,c=1$인 경우의 해를 도출하는 것도 유용하다. 이는 다음과 같이 표현할 수 있다.

$$\omega^* = \frac{V^{-1}B}{B^\prime V^{-1}B}$$

### Factor Model Misspecification

잘못된 backdoor adjustment set을 선택하면, 요인 모형이 잘못 명시되어 편향된 추정량이 발생하게 된다. 이러한 상황은 다음 두 가지 사례에서 명확히 드러난다. 첫 번째로, 연구자가 confounder를 통제하지 못한 경우를 고려하자. 이는 figure 1에서 보여주고 있다. 왼쪽에는 진짜 인과 그래프가 그려져 있고, 연구자가 이를 잘못 해석하여 오른쪽의 인과 그래프로 판단한 경우를 나타낸다. true exposure matrix는 다음과 같다.

$$B_{\text{true}} = \begin{bmatrix} \gamma_1 & \beta_1 \\ \gamma_2 & \beta_2 \end{bmatrix}$$

반면, misspecified matrix of exposures는

$$B_{\text{miss}} = \begin{bmatrix} \hat{\beta}_1 \\ \hat{\beta}_2 \end{bmatrix}$$

이다. López de Prado and Zoonekynd [2024]는 (단위 분산을 가정할 때) 자산 $n \in \{1,2\}$에 대해 $\beta_2$가 다음과 같이 추정될 것임을 보여주었다.

$$\hat{\beta}_n = \beta_n + \frac{\gamma_n \delta}{1 + \delta^2}$$

두 번째로, 연구자가 collier를 통제한 경우를 고려하자. 이는 figure 2에서 보여주고 있다. true exposure matrix는 다음과 같다.

$$B_{\text{true}} = \begin{bmatrix} \beta_1 \\ \beta_2 \end{bmatrix}$$

반면 misspecifies matrix of exposure는

$$B_{\text{miss}} = \begin{bmatrix} \hat{\theta}_1 & \hat{\beta}_1 \\ \hat{\theta}_2 & \hat{\beta}_2 \end{bmatrix}$$

이다. López de Prado와 Zoonekynd [2024]는 (단위 분산을 가정할 때) 자산 $n \in \{1,2\}$에 대해 $\beta_n$과 $\theta_n$이 다음과 같이 추정될 것임을 보여준다.

$$
\begin{align}
\beta^* =& \beta_n - \gamma_n \delta \\
\hat{\theta}_n =& \frac{\gamma_n}{1 + \gamma_n^2}
\end{align}
$$

```python
import numpy as np
import pandas as pd

N = 10000

def random(size = 1) :
    return np.random.uniform(-1.5, 1.5, size = size)

def not_controlling_for_a_confounder(delta, gamma, beta):
    Sigma = np.array([
        [ 1, delta ],
        [ delta, delta**2 + 1 ],
    ])
    Delta = np.array( [0,0] )
    B = np.vstack([gamma, beta]).T
    V = B @ Sigma @ B.T + np.diag(Delta)
    
    beta_hat = beta + gamma * delta / (1 + delta ** 2)
    w = np.linalg.solve(B.T, np.array([0,1]))
    w_mis = beta_hat / (beta_hat ** 2).sum()
    return w[0], w[1], w_mis[0], w_mis[1]
```

```python
d1 = [
    not_controlling_for_a_confounder(
        delta = random(1)[0], gamma = random(2), beta = random(2),
    )
    for _ in range(N)
]
columns = ['w1', 'w2', 'w1_mis', 'w2_mis']
d1 = pd.DataFrame(d1, columns = columns)
```

```python
def controlling_for_a_collider(delta, gamma, beta):
    B = np.vstack([np.zeros(2), beta]).T
    Delta = .1 * np.eye(2)
    V = beta.reshape((2,1)) @ beta.reshape((1,2)) + Delta
    w = np.linalg.solve(V, beta) / np.sum(beta * np.linalg.solve(V, beta))
    beta_hat = beta - gamma * delta
    theta_hat = gamma / (1 + gamma)
    B_mis = np.vstack([theta_hat, beta_hat]).T
    w_mis = np.linalg.solve(B_mis.T, np.array([0,1]))
    return w[0], w[1], w_mis[0], w_mis[1]
```

```python
d2 = [
    controlling_for_a_collider(
        delta = random(1)[0], gamma = random(2), beta = random(2),
    )
    for _ in range(N)
]
columns = ['w1', 'w2', 'w1_mis', 'w2_mis']
d2 = pd.DataFrame(d2, columns = columns)
```

### Investment Inefficiency

이전 섹션에서는 confounder를 통제하지 못한 경우와 collider를 통제한 경우 모두에서 factor에 대한 위험 귀속이 편향된다는 것을 보여주었다.이번 섹션에서는 이러한 위험 귀속의 잘못이 포트폴리오의 효율성에 미치는 영향을 연구한다. 첫 번째 경우(confounder를 통제하지 못한 경우)에서 최적 할당(optimal allocations)은 다음과 같다.

$$
\begin{align}
\omega_{\text{true}}^* = B_{\text{true}}^{\prime -1}c^\prime =& \frac{1}{\gamma_1 \beta_2 - \gamma_2 \beta_1}\begin{bmatrix}-\gamma_2 \\ \gamma_1\end{bmatrix} \\
\omega_{\text{miss}}^* =& \frac{V^{-1}B_{\text{miss}}}{B_{\text{miss}}^\prime V^{-1} B_{\text{miss}}} \\
B_{\text{miss}}^\prime =& \begin{bmatrix} \beta_1 + \frac{\gamma_1 \delta}{1 + \delta^2} & \beta_2 + \frac{\gamma_2 \delta}{1 + \delta^2}\end{bmatrix}
\end{align}
$$

```python
def get_beta_hat_failing_confounder(delta, gamma, beta):
    Sigma = np.array([
        [1, delta],
        [delta, delta**2 + 1],
    ])
    Delta = np.array([0, 0])
    B = np.vstack([gamma, beta]).T
    V = B @ Sigma @ B.T + np.diag(Delta)

    beta_hat = beta + gamma * delta / (1 + delta ** 2)
    return beta_hat[0], beta_hat[1]

N = 10000

beta1 = [
    get_beta_hat_failing_confounder(
        delta=random(1)[0], gamma=random(2), beta=random(2),
    )
    for _ in range(N)
]
columns = ['beta1_hat', 'beta2_hat']
beta1 = pd.DataFrame(beta1, columns=columns)
```

잘못 명시된 모형에서 도출된 해는 비효율적이다. 이는 투자자가 원하지 않는 위험에 노출되기 때문이다.

$$\omega_{\text{miss}}^{* \prime} B_{\text{true}} = \omega_{\text{miss}}^{* \prime} \begin{bmatrix} \gamma_1 & \beta_1 \\ \gamma_2 & \beta_2 \end{bmatrix} \neq \begin{bmatrix} 0 & 1 \end{bmatrix}$$

figure 3은 [-1.5, 1.5] 범위의 균일 분포로부터 추출한 10,000개의 매개변수 조합에 대해 이 결과를 확인한다. 두 번째 경우(collider를 통제한 경우)에서 최적 할당은 다음과 같다.

$$
\begin{align}
B_{\text{true}}^* =& \begin{bmatrix}\beta_1 \\ \beta_2\end{bmatrix} \\
\omega_{\text{true}}^* =& \frac{V^{-1}B_{\text{true}}}{B_{\text{true}}^{\prime} V^{-1} B_{\text{true}}} = 
\frac{1}{\beta_1^2 + \beta_2^2} \begin{bmatrix} \beta_1 \\ \beta_2 \end{bmatrix}
\\
\omega_{\text{miss}}^* =& B_{\text{miss}}^{\prime -1}c^\prime =
\frac{1}{\beta_2 \gamma_1 (1+\gamma_2^2) - \beta_1 \gamma_2 (1+\gamma_1^2) + \gamma_1\gamma_2 (\gamma_1^2 - \gamma_2^2) \delta}
\begin{bmatrix}
-\gamma_2(1+\gamma_1^2) \\ \gamma_1(1+\gamma_2^2)
\end{bmatrix}
\end{align}
$$

다시 한 번 말하지만, 잘못 명시된 모형에서 도출된 해는 비효율적이다. 이는 투자자가 원하지 않는 위험에 노출되기 때문이다.

$$\omega_{\text{miss}}^{* \prime} B_{\text{true}} = \omega_{\text{miss}}^{* \prime} \begin{bmatrix} \beta_1 \\ \beta_2\end{bmatrix} \neq 1$$

figure 4는 [-1.5, 1.5] 범위의 균일 분포로부터 추출한 10,000개의 parameter 조합에 대해 이 결과를 확인한다.

### Conclusion

현재의 경제계량학 정설을 따르며, 팩터 투자 연구는 모델 명세의 선택을 모델의 설명력 측면에서 정당화해 왔다. 그러나 이 과정에서 모델 변수들 간의 인과적 의존 구조를 전혀 고려하지 않았다. 이로 인해 잘못된 명세 오류가 투자 효율성이나 성과에 미치는 영향을 연구하지 않았다.

이 글은 연관성을 모델링하는 것만으로는 최적 포트폴리오를 계산하기에 충분하지 않다는 것을 보여주었다.관련 인과 그래프에 대한 지식 없이 팩터 모델을 설정하면, 모델이 잘못 명시될 가능성이 높다. 이는 위험 노출의 편향된 추정치로 이어지고, 결국 비효율적(비최적) 포트폴리오를 초래한다. 더욱이, 우리는 모델 명세 오류로 인한 편향이 너무 커서, 투자자가 기대값과 공분산을 완전히 알고 있는 경우에도 매도해야 할 자산을 매수하고, 매수해야 할 자산을 매도하는 포트폴리오를 생성할 수 있음을 보여주었다.

팩터 연구자들은 팩터 모델 명세 오류에 대한 우려를 무시해서는 안 된다. 연구자들은 **인과 탐색 분석(causal discovery analysis)**을 사용하여, 경험적 증거와 일관된 잠재적 인과 그래프를 좁혀나가는 작업을 수행해야 한다. 또한, 잠재적 confounder를 논의하고, 어떤 변수가 잠재적 collider인지 명확히 주장해야 한다. 발견된 인과 그래프를 바탕으로 연구자들은 do-calculus를 이용하여 신중하게 선택한 sufficient adjustment set을 정당화하고, 이에 따라 모델 명세를 정당화해야 한다.

인과 탐색에 대한 회의론자들은 이러한 알고리즘이 너무 부정확하여 유용하지 않다고 주장한다. 하지만 이 비판에는 세 가지 오류가 있다. 첫째, 이는 최근의 인과 그래프 추정의 발전(Glymour et al. [2019])을 간과하고 있다. 이는 투자자의 사전 관점(ex-ante views, Laudy et al. [2022])이나 사후 결과(ex-post outcomes, Sadeghi et al. [2024], Gu et al. [2023])를 반영할 수 있다. 둘째, 실무자들이 잘 알고 있듯이, 투자에서는 정확성이 중요한 것이 아니다. 미세한 개선이 손실 전략을 수익 전략으로 바꿀 수 있다. 모델 명세에서 몇 가지 confounder를 추가하거나, 몇 가지 collider를 제거하는 것만으로도 상당히 나은 전략을 달성할 수 있다. 셋째, 인과 탐색 알고리즘은 팩터 모델을 더 현실적으로 만든다. 이는 경험적 증거와 일관되지 않은 인과 그래프를 제거하여, 이론적으로 가능한 인과 그래프의 집합을 좁히기 때문이다. 이렇게 정제된 집합에서 연구자들은 모델 명세를 유도할 그래프를 선택할 수 있으며, 이 과정에서 기존에 숨겨졌던 가정을 명시적으로 드러낼 수 있다. 선택한 그래프가 여전히 틀릴 수는 있지만, 적어도 연구자는 자신의 선택을 명확히 밝히고, 그 선택을 반증 가능성(refutability)과 피드백에 노출시킬 수 있다 (López de Prado [2015, 2023a]).
